```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from torch.utils.data import Dataset, DataLoader
import torch.nn as nn
```

```python
# ============ 1. 生成数据 ============
seq_length = 300
t = np.linspace(0, 50, seq_length)
data = np.sin(t) + np.random.normal(0, 0.05, seq_length)

# ============ 2. 自定义Dataset ============
class TimeSeriesDataset(Dataset):
    def __init__(self, data, window_size):
        self.data = data
        self.window_size = window_size
        
    def __len__(self):
        return len(self.data) - self.window_size
    
    def __getitem__(self, idx):
        X = self.data[idx:idx + self.window_size]
        y = self.data[idx + self.window_size]
        # 增加特征维度: (window_size,) -> (window_size, 1)
        return torch.FloatTensor(X).unsqueeze(-1), torch.FloatTensor([y])
    
# ============ 3. 创建数据集 ============
window_size = 10
dataset = TimeSeriesDataset(data, window_size)
print(f"数据集大小: {len(dataset)}")

# 查看数据
X, y = dataset[0]
print(f"X形状: {X.shape}")  # torch.Size([10, 1])
print(f"y形状: {y.shape}")  # torch.Size([1])

# ============ 4. 划分数据集 ============
train_size = int(0.8 * len(dataset))
test_size = len(dataset) - train_size
train_dataset, test_dataset = torch.utils.data.random_split(
    dataset, [train_size, test_size]
)

train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=5, shuffle=False)
```

```python
import torch
import torch.nn as nn
class myLSTM(nn.Module):
    def __init__(self, input_size=1, hidden_size=20, output_size=1):
        super(myLSTM, self).__init__()
        self.input_size = input_size
        self.hidden_size = hidden_size
        self.output_size = output_size

        # 遗忘门
        self.bf = nn.Parameter(torch.randn(hidden_size) * 0.01)
        self.Whf = nn.Parameter(torch.randn(input_size + hidden_size, hidden_size) * 0.01)

        # 输入门
        self.bi = nn.Parameter(torch.randn(hidden_size) * 0.01)
        self.Whi = nn.Parameter(torch.randn(input_size + hidden_size, hidden_size) * 0.01)

        # 候选状态的权重
        self.bc = nn.Parameter(torch.randn(hidden_size) * 0.01)
        self.Whc = nn.Parameter(torch.randn(input_size + hidden_size, hidden_size) * 0.01)

        # 输出门
        self.bo = nn.Parameter(torch.randn(hidden_size) * 0.01)
        self.Who = nn.Parameter(torch.randn(input_size + hidden_size, hidden_size) * 0.01)

        # 输出层
        self.bot = nn.Parameter(torch.randn(output_size) * 0.01)
        self.Whot = nn.Parameter(torch.randn(hidden_size, output_size) * 0.01)

    def forward(self, x, hidden=None, ct = None):
        """
        x: (batch_size, seq_len, input_size)
        hidden: (batch_size, hidden_size)
        """
        batch_size, seq_len, _ = x.shape

        if hidden is None:
            hidden = torch.zeros(batch_size, self.hidden_size, device=x.device)
        if ct is None:
            ct = torch.zeros(batch_size, self.hidden_size, device=x.device)

        outputs = []
        for t in range(seq_len):
            xt = x[:, t, :]  # (batch, input_size)
            cat_mat = torch.cat([xt, hidden], dim=1)

            # 遗忘门
            ft = torch.sigmoid(cat_mat @ self.Whf + self.bf)

            # 输入门, 候选状态的权重
            it = torch.sigmoid(cat_mat @ self.Whi + self.bi)
            ct_hat = torch.tanh(cat_mat @ self.Whc + self.bc)
            ct = ft * ct + it * ct_hat

            # 输出门
            ot = torch.sigmoid(cat_mat @ self.Who + self.bo)
            hidden = ot * torch.tanh(ct)
            # 输出层
            outputs.append(hidden)

        # 收集所有时间步的输出
        outputs = torch.stack(outputs, dim=1)  # (batch, seq_len, hidden_size)

        # 取最后一个时间步用于输出
        last_output = outputs[:, -1, :]        # (batch, hidden_size)
        y = last_output @ self.Whot + self.bot # (batch, output_size)

        return y, outputs
```

```python
# ============ 6. 初始化模型 ============
model = myLSTM(input_size=1, hidden_size=20, output_size=1)
print("\n模型参数:")
for name, param in model.named_parameters():
    print(f"{name}: {param.shape}")

# ============ 7. 训练 ============
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

epochs = 100
train_losses = []
test_losses = []

print("\n开始训练...")
for epoch in range(epochs):
    # 训练
    model.train()
    train_loss = 0
    for batch_x, batch_y in train_loader:
        # batch_x: (batch, seq_len, 1)
        # batch_y: (batch, 1)
        pred, _ = model(batch_x)
        loss = criterion(pred, batch_y)
        
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        
        train_loss += loss.item()
    
    avg_train_loss = train_loss / len(train_loader)
    train_losses.append(avg_train_loss)
    
    # 测试
    model.eval()
    test_loss = 0
    with torch.no_grad():
        for batch_x, batch_y in test_loader:
            pred, _ = model(batch_x)
            loss = criterion(pred, batch_y)
            test_loss += loss.item()
    
    avg_test_loss = test_loss / len(test_loader)
    test_losses.append(avg_test_loss)
    
    if (epoch + 1) % 20 == 0:
        print(f'Epoch [{epoch+1}/{epochs}], '
              f'Train Loss: {avg_train_loss:.6f}, '
              f'Test Loss: {avg_test_loss:.6f}')

print("训练完成！")

# ============ 8. 可视化损失 ============
plt.figure(figsize=(12, 4))
plt.plot(train_losses, label='训练损失', linewidth=2)
plt.plot(test_losses, label='测试损失', linewidth=2)
plt.xlabel('Epoch')
plt.ylabel('MSE Loss')
plt.title('训练和测试损失曲线')
plt.legend()
plt.grid(True)
plt.show()

# ============ 9. 测试集预测 ============
model.eval()
predictions = []
targets = []

with torch.no_grad():
    for batch_x, batch_y in test_loader:
        pred, _ = model(batch_x)
        predictions.extend(pred.squeeze().numpy())
        targets.extend(batch_y.squeeze().numpy())

predictions = np.array(predictions)
targets = np.array(targets)

# 计算指标
mse = np.mean((predictions - targets) ** 2)
mae = np.mean(np.abs(predictions - targets))
rmse = np.sqrt(mse)

print(f"\n测试集评估指标:")
print(f"  MSE:  {mse:.6f}")
print(f"  RMSE: {rmse:.6f}")
print(f"  MAE:  {mae:.6f}")

# ============ 10. 可视化预测结果 ============
plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.scatter(targets, predictions, alpha=0.5, s=10)
plt.plot([targets.min(), targets.max()], 
         [targets.min(), targets.max()], 
         'r--', linewidth=2)
plt.xlabel('真实值')
plt.ylabel('预测值')
plt.title(f'预测 vs 真实 (MSE={mse:.4f})')
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(targets[:50], 'b-', label='真实值', linewidth=2)
plt.plot(predictions[:50], 'r--', label='预测值', linewidth=2)
plt.xlabel('测试样本序号')
plt.ylabel('值')
plt.title('测试集预测结果对比（前50个样本）')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
